```python
import itertools as itl
from common import plotVectors

def makeOneOvers(init):
    current = init
    while True:
        yield current
        current = 1/current

# makes no ense to have more than 2, as 1/x done 2nd time is just x again :)
sample = list(itl.islice(makeOneOvers(1+1j), 2))
plotVectors(sample, ['red'])
```

The $\frac{1}{z}$ (where $\frac{1}{z} == z^{-1}$) can be considered with $|z|e^\phi$ form, which will be  ${\frac{1}{|z|}}*e^{-\phi}$
