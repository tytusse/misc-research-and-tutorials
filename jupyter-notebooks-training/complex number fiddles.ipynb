What is the result of multiplying two "opposite" complex numbers?

```python
(-1-1j)*(1+1j)
```

Looks like the number gets stretched down to `-2j` - how to interpret that? It is a "rotation" but its not like moving vectors or something ... Fiddles below show the product is longer than components.

```python
import common
import cmath
import math
```

```python
topright=1+1j
bottomleft=(-1-1j)
a, b = topright, bottomleft
product = a*b
# Define complex numbers
vectors = [a, b, product]

common.plotVectors(vectors)
```

```python
vectors
```

```python
[abs(x) for x in vectors]
```

```python
phases =[cmath.phase(x) for x in vectors]
phases
```

```python
[math.degrees(x) for x in phases]
```

Looks like we have the following (we "Re-discovered" complex product meaning :D):

* lengths are multiplied
* phases are added 

```python
# the == wont work on raw `abx()` due to floating point precission error
round(abs(a)*abs(b), 4) == abs(product)
```

```python
cmath.phase(a) + cmath.phase(b) == cmath.phase(product)
```
